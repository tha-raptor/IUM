```python
base_url = "https://www.rottentomatoes.com/"
df_tomatoes['rotten_tomatoes_link'] = base_url + df_tomatoes['rotten_tomatoes_link'].astype(str)
df_tomatoes.head()
```

#### Normalizing data: review scores

```python
df_tomatoes['review_score'].unique()
```

```python
df_tomatoes[~df_tomatoes['review_score'].str.contains('/', na=False)]['review_score'].unique()
```

```python
letter_map = {
    'A+': 5.0, 'A': 5.0, 'A-': 4.5,
    'B+': 4.0, 'B': 3.5, 'B-': 3.0,
    'C+': 2.5, 'C': 2.0, 'C-': 1.5,
    'D+': 1.0, 'D': 1.0, 'D-': 0.5,
    'F': 0.0
}

df_tomatoes['review_score'] = df_tomatoes['review_score'].astype(str).str.strip().str.upper()
df_tomatoes['score'] = df_tomatoes['review_score'].map(letter_map)

fractions = df_tomatoes['review_score'].str.split('/', expand=True)
numerator = pd.to_numeric(fractions[0], errors='coerce')
denominator = pd.to_numeric(fractions[1], errors='coerce')
fraction_res = (numerator / denominator) * 5.0

df_tomatoes['score'] = df_tomatoes['score'].fillna(fraction_res)

df_tomatoes['score'] = df_tomatoes['score'].fillna(-1)
df_tomatoes[['review_score', 'score']]
```

```python
df_tomatoes.dtypes
```

```python
df_tomatoes.drop(columns= ['review_score'])
df_tomatoes.duplicated(subset=["movie_title", "critic_name", "publisher_name", "review_type", "review_content", "review_date"]).sum()
```

```python
df_tomatoes = df_tomatoes.drop_duplicates(subset=["movie_title", "critic_name", "publisher_name", "review_type", "review_content", "review_date"],keep="first")
df_tomatoes.shape
```
